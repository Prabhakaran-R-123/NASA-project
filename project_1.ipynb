```python
import requests
from datetime import datetime

asteroids_data = []
target = 10000
URL = f"https://api.nasa.gov/neo/rest/v1/feed?start_date=2024-01-01&end_date=2024-01-07&api_key={API_KEY}"

while len(asteroids_data) < target:
    response = requests.get(URL)
    data = response.json()
    details = data['near_earth_objects']

    for date,info in details.items():
        for i in info:
            asteroids_data.append(dict(id = int(i['id']),
                                      name = i['name'],
                                       absolute_magnitude_h = float(i['absolute_magnitude_h']),
                                       estimated_diameter_min_km = float(i['estimated_diameter']['kilometers']['estimated_diameter_min']),
                                       estimated_diameter_max_km = float(i['estimated_diameter']['kilometers']['estimated_diameter_max']),
                                       is_potentially_hazardous = i['is_potentially_hazardous_asteroid'],
                                       close_approach_date = datetime.strptime(i['close_approach_data'][0]['close_approach_date'],"%Y-%m-%d").date(),
                                       relative_velocity_kmph = float(i['close_approach_data'][0]['relative_velocity']['kilometers_per_hour']),
                                       astronomical = float(i['close_approach_data'][0]['miss_distance']['astronomical']),
                                       miss_distance_km = float(i['close_approach_data'][0]['miss_distance']['kilometers']),
                                       miss_distance_lunar = float(i['close_approach_data'][0]['miss_distance']['lunar']),
                                       orbiting_body = i['close_approach_data'][0]['orbiting_body']                
                                      ))
            if len(asteroids_data) >= target:
                break
        if len(asteroids_data) >= target:
            break
    
    URL = data['links']['next']     
```

## Inserting NASA Asterods data into SQL:

```python
import mysql.connector as db
import pandas as pd
```

```python
connection = db.connect(

    host = "localhost" ,
    user = "root",
    password = "5256",
    database = "asteroids")
```

```python
cursor = connection.cursor()
```

```python
cursor.execute("""
create table asteroids_table(
id int,
name varchar(100),
absolute_magnitude float,
estimated_diameter_min float,
estimated_diameter_max_km float,
is_potentially_hazardous_asteroid boolean
);
""")
```

```python
cursor.execute("""
create table close_approach_table(
neo_reference_id int,
close_approach_date date,
relative_velocity_kmph float ,
astronomical float,
miss_distance_km float,
miss_distance_lunar float,
orbiting_body varchar(100)
)
""")
```

```python
insert = """
insert into asteroids_table values (%s,%s,%s,%s,%s,%s)
"""
for i in asteroids_data:
    values = [
        (
            int(i['id']),
            i['name'],
            i['absolute_magnitude_h'],
            i['estimated_diameter_min_km'],
            i['estimated_diameter_max_km'],
            i['is_potentially_hazardous']
        )
    ]

cursor.executemany(insert, values)
connection.commit()
```

```python
insert = """
insert into close_approach_table values (%s, %s, %s, %s, %s, %s, %s)
"""
for i in asteroids_data:

    values = (
        int(i['id']),
        i.get('close_approach_date'),
        float(i.get('relative_velocity_kmph', 0)),
        float(i.get('astronomical', 0)),
        float(i.get('miss_distance_km', 0)),
        float(i.get('miss_distance_lunar', 0)),
        i.get('orbiting_body')
    )


    cursor.execute(insert, values)


connection.commit()
```
